```python
import sys
import os

import gradio as gr
from src.evaluate import ToxicityEvaluator

# Point to the saved models
m0_path = "results/M0_baseline/final_model"
debiased_path = "results/M2/final_model" # If you want to test M3 instead, change this to results/M3/final_model

print("Initializing Evaluator (this might take a few seconds to load both models into memory)...")
evaluator = ToxicityEvaluator(m0_path, debiased_path)
```

```python
def analyze_text(text):
    m0_prob, deb_prob = evaluator.predict(text)
    
    return {
        "Baseline Model (M0) Toxicity": m0_prob,
        "Debiased Model Toxicity": deb_prob
    }

demo = gr.Interface(
    fn=analyze_text,
    inputs=gr.Textbox(lines=3, placeholder="Enter a sentence here...", label="Input Text"),
    outputs=gr.Label(num_top_classes=2, label="Toxicity Predictions"),
    title="Toxicity Classifier: Bias Mitigation Dashboard",
    description="Compare how the Baseline model and the Debiased model score the exact same sentence. Try sentences with identity keywords!",
    examples=[
        ["I am a proud gay man."],
        ["Females are not meant to grow outside kitchen."],
        ["Are you a muslim or hindu?"],
        ["You are an absolute idiot and I hate you."],
        ["Some people in the black community feel ignored."]
    ]
)

# Setting share=True creates a public URL you can click to view the UI in a full browser tab!
demo.launch(share=True)
```
