# Exploratory Data Analysis
Calculating toxicity rate per identity vs overall base rate.

```python
from datasets import load_dataset
import pandas as pd

print("Loading dataset...")
dataset = load_dataset("google/civil_comments", split="train")
```

```python
# Sample 100k for EDA
df = dataset.shuffle(seed=42).select(range(100000)).to_pandas()

# Binarize toxicity
df['toxic'] = (df['toxicity'] >= 0.5).astype(int)

identities = [
    "male", "female", "black", "white", "muslim", 
    "christian", "jewish", "homosexual_bisexual"
]

# Calculate base toxicity rate
base_rate = df['toxic'].mean()
print(f"Overall Toxicity Base Rate: {base_rate:.2%}\n")

results = []
for identity in identities:
    # Mention means identity > 0.5
    mentioned = df[df[identity] > 0.5]
    if len(mentioned) > 0:
        tox_rate = mentioned['toxic'].mean()
        results.append({"Identity": identity, "Count": len(mentioned), "Toxicity Rate": tox_rate})

results_df = pd.DataFrame(results)
results_df
```
